```python
html_content = r"""<!DOCTYPE html>
<html lang="en">
<head>
  <meta charset="UTF-8">
  <meta name="viewport" content="width=device-width, initial-scale=1.0">
  <title>VVB Transverse Polarization Animations Matrix</title>
  
  <style>
    body {
      font-family: 'Times New Roman', Times, serif;
      background-color: #ffffff;
      color: #000000;
      margin: 20px;
      display: flex;
      flex-direction: column;
      align-items: center;
    }

    /* Rotation Notice Banner for Mobile Portrait Mode */
    .rotate-notice {
      display: none;
      background-color: #fff3cd;
      color: #856404;
      border: 1px solid #ffeeba;
      padding: 10px 16px;
      border-radius: 6px;
      margin-bottom: 15px;
      text-align: center;
      font-size: 0.95rem;
      width: 100%;
      max-width: 960px;
      box-sizing: border-box;
    }

    @media screen and (max-width: 768px) and (orientation: portrait) {
      .rotate-notice {
        display: block;
      }
    }

    .slide-container {
      width: 100%;
      max-width: 960px;
    }

    .slide-title {
      font-size: 1.25rem;
      text-align: center;
      margin-bottom: 2px;
    }

    .slide-subtitle {
      font-size: 1.05rem;
      text-align: center;
      margin-top: 0;
      margin-bottom: 16px;
    }

    .math {
      font-family: 'Cambria Math', 'STIX Two Math', 'Times New Roman', serif;
      font-style: italic;
    }

    /* Unified Control Bar Styles */
    .controls-wrapper {
      display: flex;
      align-items: center;
      justify-content: center;
      gap: 12px;
      background-color: #f2f2f2;
      border: 1px solid #ccc;
      border-radius: 6px;
      padding: 8px 16px;
      margin-top: 16px;
      width: 100%;
      box-sizing: border-box;
    }

    .play-btn {
      background-color: #0056b3;
      color: #ffffff;
      border: none;
      padding: 6px 16px;
      font-size: 1rem;
      font-family: inherit;
      border-radius: 4px;
      cursor: pointer;
      display: flex;
      align-items: center;
      justify-content: center;
      min-width: 110px;
      transition: background-color 0.2s, opacity 0.2s;
    }

    .play-btn:hover:not(:disabled) {
      filter: brightness(0.9);
    }

    .play-btn:disabled {
      background-color: #6c757d;
      cursor: not-allowed;
      opacity: 0.7;
    }

    .seek-slider {
      flex-grow: 1;
      height: 6px;
      cursor: pointer;
      accent-color: #0056b3;
    }

    .seek-slider:disabled {
      cursor: not-allowed;
      opacity: 0.5;
    }

    .time-display {
      font-size: 1rem;
      font-family: 'Times New Roman', Times, serif;
      color: #333;
      min-width: 130px;
      text-align: right;
    }

    /* Loading Status Notice Below Control Bar */
    .loading-notice {
      margin-top: 8px;
      padding: 6px 12px;
      font-size: 0.95rem;
      font-family: sans-serif;
      color: #856404;
      background-color: #fff3cd;
      border: 1px solid #ffeeba;
      border-radius: 4px;
      text-align: center;
      transition: opacity 0.5s ease;
      width: 100%;
      box-sizing: border-box;
      display: none;
    }

    .loading-notice.hidden {
      opacity: 0;
      pointer-events: none;
    }

    .animation-grid {
      display: grid;
      grid-template-columns: 70px 70px repeat(4, 1fr);
      grid-template-rows: 35px 35px repeat(3, 1fr);
      gap: 2px;
      background-color: #ffffff;
    }

    .grid-cell {
      display: flex;
      justify-content: center;
      align-items: center;
      font-size: 1.3rem;
    }

    .grey-banner {
      background-color: #a6a6a6;
      color: #000000;
      font-weight: normal;
    }

    .sigma-top {
      grid-column: 3 / span 4;
      grid-row: 1;
    }

    .sigma-left {
      grid-column: 1;
      grid-row: 3 / span 3;
    }

    .orange-cell {
      background-color: #fca385;
      color: #000000;
      font-weight: normal;
    }

    .video-cell {
      position: relative;
      width: 100%;
      aspect-ratio: 1 / 1;
      background-color: #000000;
      border: 1px solid #ffffff;
      overflow: hidden;
    }

    video {
      width: 100%;
      height: 100%;
      object-fit: cover;
      display: block;
      pointer-events: none;
      -webkit-transform: translateZ(0);
    }

    video::-webkit-media-controls {
      display: none !important;
    }
    video::-webkit-media-controls-start-playback-button {
      display: none !important;
      -webkit-appearance: none !important;
    }

    /* 右下角標記樣式（Delta ell 形式） */
    .l-overlay {
      position: absolute;
      bottom: 4px;
      right: 10px;
      color: #ffffff;
      font-family: 'Times New Roman', Times, serif;
      font-size: 1.8rem;
      pointer-events: none;
      text-shadow: 1px 1px 3px #000000, -1px -1px 3px #000000;
      z-index: 10;
      display: flex;
      align-items: center;
    }
  </style>
</head>
<body>

<div class="rotate-notice">
    <strong>Please rotate your phone for optimal display.</strong> 
</div>

<div class="slide-container">
  <div class="slide-title">
    In the below animations : <span class="math">&alpha;</span> = 0, <span class="math">&beta;</span> = <span class="math">&pi;</span>/2
  </div>
  <div class="slide-subtitle">(animations are plotted <strong>without</strong> the approximation)</div>

  <div class="animation-grid">
    <!-- Top Header: Sigma Prime -->
    <div class="grid-cell grey-banner sigma-top">
      <span class="math">&sigma;'</span> = &minus;1
    </div>

    <!-- Top Headers: Ell Prime -->
    <div class="grid-cell orange-cell" style="grid-column: 3; grid-row: 2;">
      <span class="math">&ell;'</span> = &minus;1
    </div>
    <div class="grid-cell orange-cell" style="grid-column: 4; grid-row: 2;">
      <span class="math">&ell;'</span> = 0
    </div>
    <div class="grid-cell orange-cell" style="grid-column: 5; grid-row: 2;">
      <span class="math">&ell;'</span> = 1
    </div>
    <div class="grid-cell orange-cell" style="grid-column: 6; grid-row: 2;">
      <span class="math">&ell;'</span> = 2
    </div>

    <!-- Left Header: Sigma -->
    <div class="grid-cell grey-banner sigma-left">
      <span class="math">&sigma;</span> = 1
    </div>

    <!-- Row 1: Ell = -1 -->
    <div class="grid-cell orange-cell" style="grid-column: 2; grid-row: 3;">
      <span class="math">&ell;</span> = &minus;1
    </div>
    
    <div class="video-cell" style="grid-column: 3; grid-row: 3;">
      <video src="l_-1_lp_-1.mp4#t=0.001" muted loop playsinline webkit-playsinline preload="auto" disablePictureInPicture controlsList="nodownload no-fullscreen noremoteplayback"></video>
      <!-- 第一格完整顯示 Δℓ = 0 -->
      <div class="l-overlay">&Delta;<span class="math">&ell;</span> = 0</div>
    </div>

    <div class="video-cell" style="grid-column: 4; grid-row: 3;">
      <video src="l_-1_lp_0.mp4#t=0.001" muted loop playsinline webkit-playsinline preload="auto" disablePictureInPicture controlsList="nodownload no-fullscreen noremoteplayback"></video>
      <div class="l-overlay">1</div>
    </div>

    <div class="video-cell" style="grid-column: 5; grid-row: 3;">
      <video src="l_-1_lp_1.mp4#t=0.001" muted loop playsinline webkit-playsinline preload="auto" disablePictureInPicture controlsList="nodownload no-fullscreen noremoteplayback"></video>
      <div class="l-overlay">2</div>
    </div>

    <div class="video-cell" style="grid-column: 6; grid-row: 3;">
      <video src="l_-1_lp_2.mp4#t=0.001" muted loop playsinline webkit-playsinline preload="auto" disablePictureInPicture controlsList="nodownload no-fullscreen noremoteplayback"></video>
      <div class="l-overlay">3</div>
    </div>

    <!-- Row 2: Ell = 0 -->
    <div class="grid-cell orange-cell" style="grid-column: 2; grid-row: 4;">
      <span class="math">&ell;</span> = 0
    </div>

    <div class="video-cell" style="grid-column: 3; grid-row: 4;">
      <video src="l_0_lp_-1.mp4#t=0.001" muted loop playsinline webkit-playsinline preload="auto" disablePictureInPicture controlsList="nodownload no-fullscreen noremoteplayback"></video>
      <div class="l-overlay">&minus;1</div>
    </div>

    <div class="video-cell" style="grid-column: 4; grid-row: 4;">
      <video src="l_0_lp_0.mp4#t=0.001" muted loop playsinline webkit-playsinline preload="auto" disablePictureInPicture controlsList="nodownload no-fullscreen noremoteplayback"></video>
      <div class="l-overlay">0</div>
    </div>

    <div class="video-cell" style="grid-column: 5; grid-row: 4;">
      <video src="l_0_lp_1.mp4#t=0.001" muted loop playsinline webkit-playsinline preload="auto" disablePictureInPicture controlsList="nodownload no-fullscreen noremoteplayback"></video>
      <div class="l-overlay">1</div>
    </div>

    <div class="video-cell" style="grid-column: 6; grid-row: 4;">
      <video src="l_0_lp_2.mp4#t=0.001" muted loop playsinline webkit-playsinline preload="auto" disablePictureInPicture controlsList="nodownload no-fullscreen noremoteplayback"></video>
      <div class="l-overlay">2</div>
    </div>

    <!-- Row 3: Ell = 1 -->
    <div class="grid-cell orange-cell" style="grid-column: 2; grid-row: 5;">
      <span class="math">&ell;</span> = 1
    </div>

    <div class="video-cell" style="grid-column: 3; grid-row: 5;">
      <video src="l_1_lp_-1.mp4#t=0.001" muted loop playsinline webkit-playsinline preload="auto" disablePictureInPicture controlsList="nodownload no-fullscreen noremoteplayback"></video>
      <div class="l-overlay">&minus;2</div>
    </div>

    <div class="video-cell" style="grid-column: 4; grid-row: 5;">
      <video src="l_1_lp_0.mp4#t=0.001" muted loop playsinline webkit-playsinline preload="auto" disablePictureInPicture controlsList="nodownload no-fullscreen noremoteplayback"></video>
      <div class="l-overlay">&minus;1</div>
    </div>

    <div class="video-cell" style="grid-column: 5; grid-row: 5;">
      <video src="l_1_lp_1.mp4#t=0.001" muted loop playsinline webkit-playsinline preload="auto" disablePictureInPicture controlsList="nodownload no-fullscreen noremoteplayback"></video>
      <div class="l-overlay">0</div>
    </div>

    <div class="video-cell" style="grid-column: 6; grid-row: 5;">
      <video src="l_1_lp_2.mp4#t=0.001" muted loop playsinline webkit-playsinline preload="auto" disablePictureInPicture controlsList="nodownload no-fullscreen noremoteplayback"></video>
      <div class="l-overlay">1</div>
    </div>

  </div>

  <!-- Synchronized Playback Control Bar at the Bottom -->
  <div class="controls-wrapper">
    <button id="playBtn" class="play-btn">Load Videos</button>
    <input type="range" id="seekSlider" class="seek-slider" min="0" max="6.2831853" value="0" step="0.314159" disabled>
    <div id="timeDisplay" class="time-display"><span class="math">&omega;t</span> = 0.0&pi;</div>
  </div>

  <!-- Loading Status Message Below Control Bar -->
  <div id="loadingNotice" class="loading-notice">
    ⏳ Fully buffering video data into device memory... (<span id="loadedCount">0</span>/12)
  </div>
</div>

<script>
  document.addEventListener('DOMContentLoaded', () => {
    const videos = Array.from(document.querySelectorAll('video'));
    const playBtn = document.getElementById('playBtn');
    const seekSlider = document.getElementById('seekSlider');
    const timeDisplay = document.getElementById('timeDisplay');
    const loadingNotice = document.getElementById('loadingNotice');
    const loadedCountSpan = document.getElementById('loadedCount');
    
    let isSeeking = false;
    let masterVideo = videos[0];
    const TWO_PI = Math.PI * 2;
    let isFullyLoaded = false;
    const loadedSet = new Set();

    // Initial video configurations
    videos.forEach(v => {
      v.muted = true;
      v.defaultMuted = true;
      v.pause();
    });

    // Helper: Convert time to wt (0 ~ 2pi)
    function timeToWt(currentTime, duration) {
      if (!duration) return 0;
      const progress = (currentTime % duration) / duration;
      return progress * TWO_PI;
    }

    // Helper: Convert wt to time
    function wtToTime(wt, duration) {
      if (!duration) return 0;
      return (wt / TWO_PI) * duration;
    }

    // Update display text as wt = X.X pi (1 decimal place)
    function updateWtDisplay(wt) {
      const piFraction = (wt / Math.PI).toFixed(1);
      timeDisplay.innerHTML = `<span class="math">&omega;t</span> = ${piFraction}&pi;`;
    }

    // Master Video Time Update
    masterVideo.addEventListener('timeupdate', () => {
      if (!isSeeking && masterVideo.duration && isFullyLoaded) {
        const wt = timeToWt(masterVideo.currentTime, masterVideo.duration);
        seekSlider.value = wt;
        updateWtDisplay(wt);
      }
    });

    // Track unique video loading completions
    function markVideoLoaded(index) {
      if (!loadedSet.has(index)) {
        loadedSet.add(index);
        loadedCountSpan.textContent = loadedSet.size;

        if (loadedSet.size >= videos.length) {
          onAllVideosReady();
        }
      }
    }

    function onAllVideosReady() {
      if (isFullyLoaded) return;
      isFullyLoaded = true;
      
      // Stop warm-up run & reset all videos to zero
      videos.forEach(v => {
        v.pause();
        v.playbackRate = 1.0;
        v.currentTime = 0.001;
      });

      playBtn.textContent = 'Play';
      playBtn.style.backgroundColor = '#28a745';
      playBtn.disabled = false;
      seekSlider.disabled = false;

      // Hide loading notice with smooth fade
      loadingNotice.style.opacity = '0';
      setTimeout(() => {
        loadingNotice.classList.add('hidden');
      }, 500);
    }

    // Silent Warm-up Run: Force Safari to genuinely buffer video data by active playing
    function warmUpVideo(v, idx) {
      return new Promise((resolve) => {
        let isDone = false;

        const checkBufferAndComplete = () => {
          if (isDone) return;

          // Check 1: If time reached near end, buffer is full
          const isTimeNearEnd = v.duration && (v.currentTime >= v.duration - 0.2);

          // Check 2: Inspect buffered ranges
          let isBufferedFull = false;
          if (v.buffered && v.buffered.length > 0 && v.duration) {
            const bufferedEnd = v.buffered.end(v.buffered.length - 1);
            if (bufferedEnd >= v.duration - 0.1) {
              isBufferedFull = true;
            }
          }

          if (isTimeNearEnd || isBufferedFull) {
            isDone = true;
            v.removeEventListener('timeupdate', checkBufferAndComplete);
            v.pause();
            markVideoLoaded(idx);
            resolve();
          }
        };

        v.addEventListener('timeupdate', checkBufferAndComplete);

        // High speed warm-up run
        v.playbackRate = 4.0;
        
        const promise = v.play();
        if (promise !== undefined) {
          promise.catch(() => {
            // Retry play on error
            v.play();
          });
        }
      });
    }

    // Main Control Button Click Logic
    playBtn.addEventListener('click', async () => {
      // Phase 1: User clicks "Load Videos" -> Trigger 100% full silent warm-up buffer
      if (!isFullyLoaded) {
        playBtn.disabled = true;
        playBtn.textContent = 'Buffering...';
        playBtn.style.backgroundColor = '#6c757d';
        loadingNotice.style.display = 'block';

        // Warm up all 12 videos simultaneously to populate Safari RAM buffer
        const warmUpTasks = videos.map((v, idx) => warmUpVideo(v, idx));
        await Promise.all(warmUpTasks);
        onAllVideosReady();
        return;
      }

      // Phase 2: Zero-Lag Synchronized Play / Pause
      if (masterVideo.paused) {
        try {
          const targetTime = parseFloat(seekSlider.value) === 0 ? 0.001 : wtToTime(parseFloat(seekSlider.value), masterVideo.duration);
          
          // Batch align current time
          videos.forEach(v => {
            v.currentTime = targetTime;
          });

          // Batch trigger play instantly without network wait
          const playPromises = videos.map(v => v.play());
          await Promise.all(playPromises);

          playBtn.textContent = 'Pause';
          playBtn.style.backgroundColor = '#dc3545';
        } catch (err) {
          console.error("Playback start error: ", err);
        }
      } else {
        videos.forEach(v => v.pause());
        playBtn.textContent = 'Play';
        playBtn.style.backgroundColor = '#28a745';
      }
    });

    // Slider Interaction (Seeking with 0.1pi steps)
    seekSlider.addEventListener('input', () => {
      if (!isFullyLoaded) return;
      isSeeking = true;
      const wt = parseFloat(seekSlider.value);
      const seekTime = wtToTime(wt, masterVideo.duration);
      videos.forEach(v => {
        v.currentTime = seekTime;
      });
      updateWtDisplay(wt);
    });

    seekSlider.addEventListener('change', () => {
      isSeeking = false;
    });
  });
</script>

</body>
</html>"""

with open("index.html", "w", encoding="utf-8") as f:
    f.write(html_content)

print("index.html successfully updated with Delta Ell values & iOS zero-lag video controls!")
```
